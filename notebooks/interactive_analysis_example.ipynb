# Interactive Analysis Example - Hosyana Review

This notebook demonstrates interactive line-by-line analysis of the comprehensive bibliography data.
You can run each cell individually to see results immediately, just like RStudio's "Run Line" feature.

**Usage:** 
- Click on a cell and press `Shift+Enter` to run it
- Variables persist between cells
- Results display immediately below each cell

## Setup and Data Loading

```python
# Import libraries (run this cell first)
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import json
from collections import Counter

# Set up plotting
plt.style.use('default')
sns.set_palette("husl")

print("✓ Libraries loaded successfully")
```

```python
# Load the comprehensive data
df = pd.read_csv('FINAL_methods_analysis.csv')
print(f"Loaded {len(df)} papers")
print(f"Columns: {list(df.columns[:10])}...")  # Show first 10 columns
```

```python
# Quick data overview
df.head(3)  # This will display first 3 rows in a nice table
```

## Interactive Data Exploration

```python
# Check data quality - run this line to see missing data
missing_data = df.isnull().sum()
print("Missing data by column:")
print(missing_data[missing_data > 0])  # Only show columns with missing data
```

```python
# Year distribution - run this to see publication years
year_counts = df['YEAR'].value_counts().sort_index()
print(f"Year range: {df['YEAR'].min()} - {df['YEAR'].max()}")
print(f"Most productive year: {year_counts.idxmax()} with {year_counts.max()} papers")
```

```python
# Method detection summary - run this line by line to explore
papers_with_methods = df[df['HAS_METHODS'] == 1]
detection_rate = len(papers_with_methods) / len(df) * 100

print(f"Papers with methods detected: {len(papers_with_methods)}")
print(f"Detection rate: {detection_rate:.1f}%")
```

## Method Analysis (Interactive)

```python
# Get method columns
method_columns = [
    'DECISION_ANALYSIS', 'POLICY_INTERVENTION', 'UNCERTAINTY_ANALYSIS',
    'STAKEHOLDER_EXPERT', 'MODELING_SIMULATION', 'BAYESIAN_PROBABILISTIC',
    'COMPUTER_ASSISTED', 'VALUE_INFORMATION', 'MULTI_CRITERIA', 'OPTIMIZATION',
    'ECONOMIC_EVALUATION', 'GAME_THEORY', 'BEHAVIORAL_PSYCHOLOGY',
    'SYSTEMS_COMPLEXITY', 'TECHNOLOGY_INNOVATION', 'ENVIRONMENTAL_SUSTAINABILITY',
    'QUALITY_PERFORMANCE', 'RISK_SAFETY', 'FORECASTING_PREDICTION', 'EVALUATION_ASSESSMENT'
]

print(f"Found {len(method_columns)} method categories")
```

```python
# Calculate method popularity (run this to see top methods)
method_counts = df[method_columns].sum().sort_values(ascending=False)
print("Top 10 methods:")
for method, count in method_counts.head(10).items():
    clean_name = method.replace('_', ' ').title()
    print(f"  {clean_name}: {count} papers")
```

```python
# Quick visualization of top methods
plt.figure(figsize=(12, 6))
top_methods = method_counts.head(8)
clean_names = [name.replace('_', ' ').title() for name in top_methods.index]

plt.bar(range(len(top_methods)), top_methods.values)
plt.xticks(range(len(top_methods)), clean_names, rotation=45, ha='right')
plt.title('Top 8 Decision Support Methods')
plt.ylabel('Number of Papers')
plt.tight_layout()
plt.show()
```

## Temporal Analysis (Decade Trends)

```python
# Create decade column
df['decade'] = (df['YEAR'] // 10) * 10
df['decade'] = df['decade'].fillna(2020)  # Handle missing years

# Look at decade distribution
decade_counts = df['decade'].value_counts().sort_index()
print("Papers by decade:")
for decade, count in decade_counts.items():
    print(f"  {int(decade)}s: {count} papers")
```

```python
# Focus on specific decade (change the decade here)
decade_to_analyze = 2010
decade_data = df[df['decade'] == decade_to_analyze]

print(f"\n{decade_to_analyze}s Analysis:")
print(f"Total papers: {len(decade_data)}")
print(f"Papers with methods: {len(decade_data[decade_data['HAS_METHODS'] == 1])}")
```

```python
# Top methods for this decade
decade_methods = decade_data[method_columns].sum().sort_values(ascending=False)
print(f"\nTop 5 methods in {decade_to_analyze}s:")
for method, count in decade_methods.head(5).items():
    clean_name = method.replace('_', ' ').title()
    print(f"  {clean_name}: {count} papers")
```

## Recent Trends (2020-2024)

```python
# Focus on recent years (2020-2024)
recent_data = df[(df['YEAR'] >= 2020) & (df['YEAR'] <= 2024)]
print(f"Recent papers (2020-2024): {len(recent_data)}")

# Year by year breakdown
recent_by_year = recent_data['YEAR'].value_counts().sort_index()
print("\nYear-by-year breakdown:")
for year, count in recent_by_year.items():
    print(f"  {int(year)}: {count} papers")
```

```python
# Top methods in recent period
recent_methods = recent_data[method_columns].sum().sort_values(ascending=False)
print("Top methods in 2020-2024:")
for method, count in recent_methods.head(8).items():
    clean_name = method.replace('_', ' ').title()
    percentage = count / len(recent_data[recent_data['HAS_METHODS'] == 1]) * 100
    print(f"  {clean_name}: {count} papers ({percentage:.1f}%)")
```

## Custom Analysis (Edit and Run)

```python
# Example: Find papers using multiple methods
df['method_count'] = df[method_columns].sum(axis=1)
multi_method = df[df['method_count'] > 2]

print(f"Papers using 3+ methods: {len(multi_method)}")
print(f"Average methods per paper: {df['method_count'].mean():.2f}")
```

```python
# Example: Search for specific terms in titles
search_term = "decision"  # Change this to search for different terms
matching_papers = df[df['TITLE'].str.contains(search_term, case=False, na=False)]

print(f"Papers with '{search_term}' in title: {len(matching_papers)}")
if len(matching_papers) > 0:
    print("\nExample titles:")
    for title in matching_papers['TITLE'].head(3):
        print(f"  - {title}")
```

```python
# Example: Create a simple trend plot
plt.figure(figsize=(10, 6))

# Choose a method to track over time
method_to_track = 'DECISION_ANALYSIS'  # Change this to track different methods

trend_data = df.groupby('decade')[method_to_track].sum()
trend_data = trend_data[trend_data.index >= 1970]  # From 1970 onwards

plt.plot(trend_data.index, trend_data.values, marker='o', linewidth=2, markersize=8)
plt.title(f'Trend: {method_to_track.replace("_", " ").title()} Over Time')
plt.xlabel('Decade')
plt.ylabel('Number of Papers')
plt.grid(True, alpha=0.3)
plt.show()
```
